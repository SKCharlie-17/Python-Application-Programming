```python
#------------------question 5-----------------

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("CALCULATOR")

op = input("enter the operator  + - * /  : " )

if op == "+":
    print(a+b)

elif op == "-":
    print(a-b)

elif op == "*":
    print(a*b)

elif op == "/":
    print(a/b)

else:
    print("select an operator!!!!")
```
