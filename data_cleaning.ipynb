```python
def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            zipper = strings[i] + strings[i + 1]
            file.write(zipper + "\n")

zipper_file("text.txt", ["Hello", "World", "Good", "Morning"])
```

```python
def reformat_date(dates):
    months = [
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ]

    result = []

    for date in dates:
        month, day, year = date.split("/")
        month_name = months[int(month) - 1]

        sentence = "Office Hours on " + month_name + " " + day + ", " + year
        result.append(sentence)

    return result


dates = ["01/15/2025", "12/09/2024", "03/21/2026"]
print(reformat_date(dates))
```
