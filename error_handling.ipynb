```python
def safe_divide(numerator, denominator):

    if denominator == 0:
        print("Error: Cannot divide by Zero.")
        return None

    divide = numerator / denominator
    return divide

n = float(input("Enter the numerator: "))
d = float(input("Enter the denominator: "))

print(safe_divide(n, d))
```

```python
import math

def square_root(number1):

    if number1 < 0:
        print("Invalid input! Please enter a positive integer or a float value")
        return None
    root = math.sqrt(number1)

    return root

number = float(input("Enter a number to find the square root: "))

print(square_root(number))
```

```python
def calc(num):
    try:
        result = num / (num-5)
        return result
    except Exception as e:
        print("An error occurred during calculation.")
        return None

num = float(input("Enter a number to calculate: "))
print(calc(num))
```
