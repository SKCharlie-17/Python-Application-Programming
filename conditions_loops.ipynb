```python
for player in players:
    if player['achievements'] > 10:
        print(player["name"] , player["sport"],player["achievements"])
    else: 
        print(player['name'] + " has less than 10 achievements.")
```

```python
num = int(input("Enter a number between 1 and 9999: "))

if num%2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
word = str(input("Enter a word: "))
l = len(word) - 1
rev = ""
for i in range(len(word)):
    rev += word[l]
    l -= 1
if rev == word:
    print("The word is a palindrome.")

else:
    print("The word is not a palindrome.")
```

```python
num = str(input("enter a number : "))
l = len(num) - 1
sum_1 = 0
while l >= 0:
    sum_1 += int(num[l])
    l -= 1



if int(num)%int(sum_1) == 0:
    print("the number is divisible by the sum of its digits")

else:
    print("it is not divible")
```
